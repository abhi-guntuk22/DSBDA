```python
import pandas as pd

import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv('/content/CAR.csv')
```

```python
df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df=df.dropna()
```

```python
plt.title("Selling Price Distribution")
sns.histplot(x=df['selling_price'],bins=20,kde=True)
plt.show()
```

1. Right-skewed distribution (positive skew)
Most values are concentrated on the left, with a long tail to the right.

Meaning:

Most cars are low to moderately priced
A few cars are very expensive

In your graph:
Most cars seem around:

₹0 to ₹10 lakh (mostly much lower end)

with a few going up to:

₹80–90 lakh

```python
sns.boxplot(x=df['selling_price'])
plt.show()
```

```python
sns.histplot(df['km_driven'], kde=True)
plt.title("KM Driven Distribution")
plt.show()
```

```python
sns.countplot(x=df['fuel'], data=df)
plt.title("Fuel Type Distribution")
plt.show()
```

```python
sns.scatterplot(x=df['km_driven'],y=df['selling_price'])
```

```python
sns.violinplot(x=df['fuel'],y=df['selling_price'],data=df)
```

```python
sns.catplot(x='fuel',y='selling_price',data=df,kind='box')
```

```python
sns.heatmap(df.corr(numeric_only=True),annot=True)
```

```python
plt.figure(figsize=(12,6))
sns.barplot(x=df['year'],y=df['selling_price'],hue=df['fuel'])
```
