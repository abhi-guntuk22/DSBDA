```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("glass.csv")
```

```python
sns.pairplot(df)
plt.show()


# Pairplot shows relationships between multiple variables.
# It helps visualize clusters of different glass types.
```

```python
plt.figure(figsize=(6,4))
plt.hist(df['RI'], bins=20)
plt.title("Distribution of Refractive Index (RI)")
plt.xlabel("RI")
plt.ylabel("Frequency")
plt.show()
# Histogram shows the distribution of RI values.
# We can see how frequently each RI range occurs.
# Most glass samples fall in a specific RI range.
```

```python
plt.figure(figsize=(6,4))
sns.countplot(x='Type', data=df)
plt.title("Count of Different Glass Types")
plt.show()

# This graph shows how many samples belong to each glass type.
# It helps identify if the dataset is balanced or imbalanced.
```

```python
plt.figure(figsize=(6,4))
plt.scatter(df['Na'], df['Mg'])
plt.title("Scatter Plot of Na vs Mg")
plt.xlabel("Sodium (Na)")
plt.ylabel("Magnesium (Mg)")
plt.show()

# Scatter plot shows the relationship between Na and Mg.
# Each point represents one glass sample.
# It helps identify correlation or clustering patterns.
```

```python
plt.figure(figsize=(6,4))
sns.boxplot(x='Type', y='Al', data=df)
plt.title("Boxplot of Aluminum (Al) by Glass Type")
plt.show()

# Boxplot helps detect outliers and spread of Aluminum values
# across different glass types.
```

```python
plt.figure(figsize=(10,6))
corr = df.corr()

sns.heatmap(corr, annot=True)
plt.title("Correlation Heatmap")
plt.show()

# Heatmap shows correlation between all variables.
# Values close to 1 indicate strong positive correlation,
# values close to -1 indicate strong negative correlation.
```

```python
plt.figure(figsize=(6,4))
sns.barplot(x='Type', y='Na', data=df)
plt.title("Average Sodium (Na) per Glass Type")
plt.show()

# Bar plot shows average Sodium value for each glass type.
# Useful for comparing mean values between categories.
```
